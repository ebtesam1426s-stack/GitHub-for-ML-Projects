## Section 1 — Image Sampling and Quantization

- **Sampling** = how many pixels we keep (spatial resolution). نصغر الصور نقلل البكسلز
- **Quantization** = how many gray levels we keep (intensity resolution).

```python
def sample_image(image, factor):
    """Downsamples the image by the given factor."""
    height, width = image.shape[:2] # ابي اصغرها للنص
    sampled_image = cv2.resize(image, (width // factor, height // factor), interpolation=cv2.INTER_NEAREST) #نطبق العمليه ونقسم الهاي والودث على اثنين و بياخذ اقرب ال element in array
    return sampled_image


def quantize_image(image, levels):
    """Reduces the number of grayscale levels in the image.""" #  نقربها من الصفر وتصير دارك
    quantized_image = np.floor(image / (256 // levels)) * (256 // levels) #  (عدد الدلرجات)الليفلز كامله على عدد الليفلز اللي احنا مختارينه
    quantized_image = quantized_image.astype(np.uint8)
    return quantized_image


def plot_images(original, sampled, quantized):
    plt.figure(figsize=(12, 4)) #عرض
    plt.subplot(1, 3, 1), plt.imshow(original, cmap='gray')
    plt.title('Original Image'), plt.axis('off') # off
    plt.subplot(1, 3, 2), plt.imshow(sampled, cmap='gray')
    plt.title('Sampled Image'), plt.axis('off')
    plt.subplot(1, 3, 3), plt.imshow(quantized, cmap='gray')
    plt.title('Quantized Image'), plt.axis('off')
    plt.show()
```

```python
original_image = data.camera() # from dataset

sampling_factor = 14 #divid it by 14
quantization_levels = 9

sampled_image = sample_image(original_image, sampling_factor)
quantized_image = quantize_image(original_image, quantization_levels)

plot_images(original_image, sampled_image, quantized_image)
```

## Section 2 — Arithmetic Operations

Add two images by converting them to numpy arrays and adding element-wise. لانها اراي اقدر اجمعهم على بعض

```python
img1 = Image.fromarray(data.camera()) # 1- نعرفهم ونجيب الصور الي نبيهم من الداتا سيت
img2 = Image.fromarray(data.moon())

new_size = (400, 400)
img1 = img1.resize(new_size, Image.Resampling.LANCZOS)# حولناها الي image  عشان نقدر نستخدم resize
img2 = img2.resize(new_size, Image.Resampling.LANCZOS) # يخختار البكسل الاقرب عشان يحطها في الصوره الثانيه ويسوي عمليات حسابيه اسمها مينيوبليشن

im1arr = np.asarray(img1)
im2arr = np.asarray(img2) # رجعناها الى array عشان نجري عليها عمليات حسابيه

addition = im1arr + im2arr
result_image = Image.fromarray(addition) # نرجع نحولها الى image

plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1), plt.imshow(img1, cmap='gray'), plt.title('Image 1'), plt.axis('off')
plt.subplot(1, 3, 2), plt.imshow(img2, cmap='gray'), plt.title('Image 2'), plt.axis('off')
plt.subplot(1, 3, 3), plt.imshow(result_image, cmap='gray'), plt.title('Image 1 + Image 2'), plt.axis('off')
plt.show()
```

## Section 3 — Set / Logical Operations

Pixel-wise bitwise operations on two images. Lab demo: **union (OR)**. #لانها عباره عن اراي


```python
img3 = Image.fromarray(data.horse())
img4 = Image.fromarray(data.binary_blobs(length=400))

img3 = img3.resize(new_size, Image.Resampling.LANCZOS)
img4 = img4.resize(new_size, Image.Resampling.LANCZOS)

im3arr = np.asarray(img3)
im4arr = np.asarray(img4)

union = im4arr | im3arr
result_image_2 = Image.fromarray(union)

plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1), plt.imshow(img3, cmap='gray'), plt.title('Image A'), plt.axis('off')
plt.subplot(1, 3, 2), plt.imshow(img4, cmap='gray'), plt.title('Image B'), plt.axis('off')
plt.subplot(1, 3, 3), plt.imshow(result_image_2, cmap='gray'), plt.title('Union (A | B)'), plt.axis('off')
plt.show()
```

## Task 1 — Change sampling and quantization parameters and observe the effects

Try multiple values of the sampling factor and the number of quantization levels to see how each affects the image.

### 1a) Effect of sampling factor (spatial resolution)

```python


def sample_image(image, factor):
    h, w = image.shape[:2]
    sampled = cv2.resize(image, (w // factor, h // factor), interpolation=cv2.INTER_NEAREST)
    return sampled

original = data.camera()

factors = [2, 4, 8, 14]

plt.figure(figsize=(15, 8))

plt.subplot(2, 3, 1)
plt.imshow(original, cmap='gray')
plt.title("Original")
plt.axis('off')

for i, f in enumerate(factors):
    sampled = sample_image(original, f)
    plt.subplot(2, 3, i+2)
    plt.imshow(sampled, cmap='gray')
    plt.title(f"Sampling Factor = {f}")
    plt.axis('off')

plt.show()
```

### 1b) Effect of quantization levels (intensity resolution)

```python


def quantize_image(image, levels):
    step = 256 // levels
    quantized = (image // step) * step
    return quantized.astype(np.uint8)
original = data.camera()

levels_list = [2, 4, 8, 16, 32, 64, 128]

plt.figure(figsize=(15, 10))

plt.subplot(3, 3, 1)
plt.imshow(original, cmap='gray')
plt.title("Original")
plt.axis('off')

for i, lvl in enumerate(levels_list):
    quantized = quantize_image(original, lvl)
    plt.subplot(3, 3, i+2)
    plt.imshow(quantized, cmap='gray')
    plt.title(f"Levels = {lvl}")
    plt.axis('off')

plt.show()
```

## Task 2 — Image Arithmetic and Set Operations

Required operations:
1. Subtract two images and display the result
2. Add one image with a constant value 175
3. Set difference on two grayscale images
4. Symmetric difference on two grayscale images
5. Intersection on two grayscale images

### 2a) Subtract two images

```python

imgA = data.camera()
imgB = data.moon()

imgA = imgA[:400, :400]
imgB = imgB[:400, :400]

subtracted = imgA.astype(np.int16) - imgB.astype(np.int16)
subtracted = np.clip(subtracted, 0, 255).astype(np.uint8)

plt.figure(figsize=(10,5))
plt.subplot(1,2,1); plt.imshow(imgA, cmap='gray'); plt.title("Image A"); plt.axis('off')
plt.subplot(1,2,2); plt.imshow(subtracted, cmap='gray'); plt.title("A - B (Subtraction)"); plt.axis('off')
plt.show()
```

### 2b) Add image with constant value 175

```python

img = data.camera()

added_constant = img + 175
added_constant = np.clip(added_constant, 0, 255).astype(np.uint8)

plt.figure(figsize=(10,5))
plt.subplot(1,2,1); plt.imshow(img, cmap='gray'); plt.title("Original Image"); plt.axis('off')
plt.subplot(1,2,2); plt.imshow(added_constant, cmap='gray'); plt.title("Image + 175"); plt.axis('off')
plt.show()
```

### 2c-2e) Set operations on grayscale images

| Operation | Bitwise formula | Meaning |
|---|---|---|
| **Intersection** | `A & B` | bits set in both |
| **Set difference** | `A & ~B` (A − B) | bits in A but not in B |
| **Symmetric difference** | `A ^ B` | bits in A or B but not both |
| **Union** (extra) | `A \| B` | bits in either |

```python
imgA = data.camera()
imgB = data.moon()

imgA = imgA[:400, :400]
imgB = imgB[:400, :400]

intersection = imgA & imgB

set_difference = imgA & (~imgB)

symmetric_diff = imgA ^ imgB

union = imgA | imgB

plt.figure(figsize=(15,10))

plt.subplot(2,3,1); plt.imshow(imgA, cmap='gray'); plt.title("Image A"); plt.axis('off')
plt.subplot(2,3,2); plt.imshow(imgB, cmap='gray'); plt.title("Image B"); plt.axis('off')

plt.subplot(2,3,3); plt.imshow(intersection, cmap='gray'); plt.title("Intersection (A & B)"); plt.axis('off')
plt.subplot(2,3,4); plt.imshow(set_difference, cmap='gray'); plt.title("Set Difference (A & ~B)"); plt.axis('off')
plt.subplot(2,3,5); plt.imshow(symmetric_diff, cmap='gray'); plt.title("Symmetric Difference (A ^ B)"); plt.axis('off')
plt.subplot(2,3,6); plt.imshow(union, cmap='gray'); plt.title("Union (A | B)"); plt.axis('off')

plt.show()
```
